```python
import joblib
import os
import pandas as pd
from pathlib import Path
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
from sklearn.neighbors import KNeighborsClassifier as KNN
from sklearn.tree import DecisionTreeClassifier
```

```python
TRAINING_DATA_PATH = Path(os.getcwd()).parent / "data/processed/fraud_train.csv"
TESTING_DATA_PATH = Path(os.getcwd()).parent / "data/processed/fraud_test.csv"
MODEL_FOLDER_PATH = Path(os.getcwd()).parent / "models/"
```

```python
# Load the training data

training_df = pd.read_csv(TRAINING_DATA_PATH)
testing_df = pd.read_csv(TESTING_DATA_PATH)

labels = ["is_fraud"]
features = [heading for heading in training_df.columns if heading not in labels]
```

```python
# Train the models

N_NEIGHBORS = 6
RANDOM_STATE = 42
MAX_DEPTH = 5

knn = KNN(n_neighbors=N_NEIGHBORS)
knn.fit(training_df[features], training_df[labels].values.ravel())

decisionTree = DecisionTreeClassifier(random_state=RANDOM_STATE, max_depth=MAX_DEPTH)
decisionTree.fit(training_df[features], training_df[labels])
```

```python
# Evaluate the models

def evaluate_model(labels, predictions):
    accuracy = accuracy_score(labels, predictions)
    precision = precision_score(labels, predictions)
    recall = recall_score(labels, predictions)
    f1 = f1_score(labels, predictions)
    return {"accuracy": accuracy, "precision": precision, "recall": recall, "f1": f1}


knn_eval = evaluate_model(testing_df[labels], knn.predict(testing_df[features]))
decisionTree_eval = evaluate_model(testing_df[labels].values.ravel(), decisionTree.predict(testing_df[features]))

print(f"KNN Evaluation: {knn_eval}")
print(f"Decision Tree Evaluation: {decisionTree_eval}")
```

```python
# Save models

joblib.dump(knn, MODEL_FOLDER_PATH / "knn.pkl")
joblib.dump(decisionTree, MODEL_FOLDER_PATH / "decision_tree.pkl")
```
